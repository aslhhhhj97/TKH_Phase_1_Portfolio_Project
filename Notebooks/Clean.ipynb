```python
df.drop(columns=['Surface_Area'], inplace=True)
```

```python
dfd = pd.get_dummies(df, columns=['Orientation'],prefix="Ori", drop_first=True)
dfd = pd.get_dummies(dfd, columns=['Glazing_Area'],prefix="GA", drop_first=True)
dfd = pd.get_dummies(dfd, columns=['Glazing_Distribution'],prefix="GD",drop_first=True)
```

Note that name for the dummy columns are shortened.
Ori-Orintation
GA-GlazingArea
GD-GlazingDistribution

```python
df7=dfd[dfd['Overall_Height']==7]
df35=dfd[dfd['Overall_Height']==3.5]
```

Created new dataframes, seperated by the height

```python
df7.drop('Overall_Height', axis=1, inplace=True)
df35.drop('Overall_Height', axis=1, inplace=True)
```

Rmoved the height columns since now the data is seperated by the height
